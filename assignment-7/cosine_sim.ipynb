**Cosine Similarity**


creating text we want to vectorize

```python
text1 = "I love playing basketball"
text2 = "I love playing basketball and soccer"
compare_list = [text1,text2]
```

vectorizing the text

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()

#vectorize the texts
tf_idf_matrix = tfidf_vectorizer.fit_transform(compare_list)
```

View matrix and assigned values

```python
tfidf_vectorizer.vocabulary_
```

view the vectors


```python
print(tf_idf_matrix)
```

calculate cosine similarity between the text (which has now been vectorized)


```python
from sklearn.metrics.pairwise import cosine_similarity
cos_sim = cosine_similarity(tf_idf_matrix)
print(f"The cosine similarity between the two tweets is {cos_sim[0,1]}")
```
